## Union

```python
from typing import Union

def square(x:Union[float,int])->float:
    return x*x

x=5 ## this is fine because x is an integer

print(square(x))

x=1.234 ## this is also fine

print(square(x))


## Union lets you say that a value can be more than one type
## Type saftey as it can provide hints to help catch incorrect usage
## flexible and easy to code

```

## Optional

```python
from typing import Optional


def nice_message(name:Optional[str])->None:
    if name is None:
        print("hey random person")

    else:
        print(f"Hi there,{name}")
```

```python
nice_message('Bob')
```

```python
nice_message(None)
```

```python
nice_message()
```

```python
## it can be either string or None value
```

## Sequence

Sequence is a type hint in Python that means:

"Any ordered collection of items."

It doesn't matter whether it's a list, tuple, or string. As long as the items have an order (first, second, third...), it's a Sequence.

```python
## Sequence- To automatically handle the state updates for sequences such as by adding new messages to a chat history
from typing import Sequence

def print_numbers(numbers:Sequence[int]):
    for num in numbers:
        print(num)

print_numbers([1,2,3,4])
```

```python
print_numbers((1,2,3,4))
```
