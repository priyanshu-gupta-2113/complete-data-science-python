## Technique: Suffix-Based Feature Grouping

Suffix-Based Feature Grouping is an EDA and feature engineering technique used to organize columns that share the same suffix into separate groups or DataFrames. This is particularly useful when column names follow a naming convention, such as measurements from different sensors, time points, or statistics.

Instead of manually selecting columns, we can automatically group them based on their suffix.

```python
import pandas as pd

df = pd.DataFrame({
    "temp_A": [100, 102, 98],
    "pressure_A": [40, 42, 39],
    "speed_A": [10, 11, 12],

    "temp_B": [95, 97, 96],
    "pressure_B": [35, 36, 37],
    "speed_B": [8, 9, 10],

    "Defective_flag": [0, 1, 0]
})
df
```

```python
from collections import defaultdict
suffix_groups=defaultdict(list)

for col in df.columns:
    if "_" in col:
        suffix=col.split('_')[-1]
        suffix_groups[suffix].append(col)
suffix_groups
```

```python
df_suffix={}


for suffix,cols in suffix_groups.items():
    df_suffix[suffix]=df[cols]

df_suffix
```

```python
df_suffix['A']
```

```python
df_suffix['B']
```
