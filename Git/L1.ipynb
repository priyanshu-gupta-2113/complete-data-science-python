***Version Control System (VCS) / Source Control Manager (SCM)***

When you create a software, you first release its very basic version , then you update it and release new version. They make the software version by version. In the whole process, the software you use to control all the versions is called Version Control system.

Formal Definition:

Version Control system is a software tool that tracks, records and manages changes to files and source code over time. It acts like a time machine for your project, letting you see who made changes, revert to older working versions, and work with a team without losing data

## Types of VCS:
1. Centralised
2. Distributed


Centralised: Source code of the software is put in one machine (central machine) and all other machines constantly communicates with central main machine in order to make changes

Distributed: Source code of the software is available in all machines. 

## Advantages of VCS:

1. Version Control: You are able to maintain versions (different snapshots). You can restore to previous state always

2. Bug Fixing: You just write code and code and eventually you face a bug. You don't know where the bug is, You can compare different states and identify where the bug lies 

3. Doing non-linear development: Earlier we used to add features sequentailly (linearly), By VCS we can develop features parallely. (through Branching and Merging)

4. Collaborative Development: Helps in working as a team.

What is a Commit ?

A commit is like taking a photo (snapshot) of your project at a certain moment of time. Every time you make a commit, you say: I want to save this version of code so I can go back to it later.


What does Commit include ?

Commit stores:
- Exact state of your files
- Message describing what changed
- Unique ID (called SHA)

## Commiting stage

Meaning: Save a permanent snapshot of the code right now. You must give a message so future you know what changed.

## When to Commit ?

It is not like you just write 2 Lines of code and you do commit. When you achieve something substantial then only you commit. Whenever you are done with an independent feature then you commit.

## Creating diff versions of a software

Whenever you download a software , we also see V1.1.0 or like V2.1.4 ... These are called version

Synatx : X.Y.Z      Eg- V1.2.3

V 1.2.3 -> V 2.2.3 : Major changes are made 

1. X -> The major version, used for making major and backward incompatible changes... Means if we go from 1 to 2 then it is a very major change and if a person uses 2, he won't use 1


Suppose we use Python 3.11.2, 3.11.2 is not compatible with Python 2.Y.Z ..... a user can only uses all libraries related to version 3 only , other libraries of different versions are not compatibe


2. Y -> The minor version, used for adding functionality while maintaining backwards compatibility..

V 1.1 -> V 1.2 : You add some new feature in your new version

3. Z -> The patch version, used for making small bug fixes while maintaining backward compatibility.


Suppose we released 1.1.5 but 1.1.5 has some errors , so we revert back from 1.1.5 to 1.1.4 to correct errors
