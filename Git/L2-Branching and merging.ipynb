## The problem of Conflict (Merge Conflict)

A Git conflict (specifically a merge conflict) is an event that occurs when Git is unable to automatically combine changes from two different commits or branches. While Git is usually excellent at merging code automatically, it pauses and flags a conflict when competing changes happen in the exact same place, requiring a human to manually choose which version to keep.

Why Conflicts Happen?
- Same line changed differently: Two developers edit the exact same line of code in different branches, and then try to merge them.
- File deleted vs. file edited: One person deletes a file in their branch while another person updates the content of that same file.
- File renamed vs. file edited: One branch changes the file name while another branch modifies its code.

## What a Conflict Looks Like ? 

When a conflict occurs, Git pauses the merge process and injects conflict markers directly into the affected file:

```python
# <<<<<<< HEAD
# This is the code on your current active branch.
# =======
# This is the incoming code from the branch you are trying to merge.
# >>>>>>> feature-branch
# Use code with caution.



## <<<<<<< HEAD: Indicates the start of the changes on your current branch. Everything below this line(until the next indicator) is code of current branch
## =======: The divider separating the two conflicting versions. Is the end of original lines, everthing that follows (until the next indicator) is what is on the branch that is being merged in.
## >>>>>>> branch-name: Indicates the end of the conflicting text from the incoming branch. Is the ending indicator of what is on the branch that is being merged in.
```
