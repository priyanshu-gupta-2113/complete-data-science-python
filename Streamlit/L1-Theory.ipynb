## What is Streamlit ?

Model should run on a website or on an app. We deploy model on a website, problem is when you deploy model on website, you need web development(HTML,CSS,Javascript).

Streamlit simplifies this process of web dev. You can create websites without needing to know HTML,CSS etc

Streamlit is a plugin tool by which data scientist without even knowing concepts of HTML can build data application

## Main benefits of Streamlit

1. Build websites without needing to know Web development concepts

2. If we do analysis on a data by pandas and we can convert it into website

3. We can convert model to websites



## Technical Side of Streamlit

Streamlit library is a library built on top of Javascript library called React (Frontend library).

Streamlit uses React for frontend and for backend it uses Python



## Disadvantages of Streamlit:

- We cannot do too much customization


## To run the file of Streamlit

streamlit run "file_name.py"



You get global and local URL

You don't need to reload the website to see the changes
