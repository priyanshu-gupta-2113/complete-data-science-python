## Syntax:

## make_classification()

What is it?

It creates an artificial dataset for a classification problem.

Suppose we want to predict

Student-> Pass or Fail

Instead of collecting student data, sklearn creates fake students.

```python
from sklearn.datasets import make_classification
X,y=make_classification(
     n_samples=10,
     n_features=2,
     n_classes=2,
     random_state=42,
     n_redundant=0,
     n_informative=2
)
```

```python
X
```

## Advantages:

No need to download data.

Perfect for learning ML.

Control dataset complexity.

Add noise whenever required.

Test algorithms quickly.

Useful in interviews and research

## make_blobs()

What is it?

## make_blobs() generates clusters (groups of points).

It is mainly used to learn and test clustering algorithms like:

K-Means
DBSCAN
Mean Shift

Unlike make_classification(), it doesn't focus on complex decision boundaries. Instead, it creates well-separated groups

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

X,y=make_blobs(
    n_samples=200, ## 200 data points
    centers=3, ## 3 clusters .. each center becomes 1 cluster
    random_state=42, 
    n_features=2,
    cluster_std=0.8 ## Controls how spread out each cluster is. Higher value = more overlap between clusters.
)
```

```python
plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Three Clusters")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

## make_gaussian_quantiles()

What is it?

Instead of creating separate clusters,it creates One Big Gaussian Cloud and divides it into classes using circles (quantiles).

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_gaussian_quantiles

X,y=make_gaussian_quantiles(
    n_features=2,
    n_classes=3,
    random_state=42
)

plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Gaussian Quantiles")
plt.show()
```

## make_circles()

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles

X,y=make_circles(
    noise=0.2,
    factor=0.1,
    random_state=42,)

plt.scatter(X[:,0],X[:,1],c=y)
```
