## Train Test Split

## Numpy Arrays to pytorch

```python
import torch
X_train=torch.from_numpy(X_train)
X_test=torch.from_numpy(X_test)
y_train=torch.from_numpy(y_train)
y_test=torch.from_numpy(y_test)
```

## Creating and defining a neural network

```python
## no of weights= no of input features
class MySimpleNN():

  def __init__(self,X):
    self.weights=torch.rand(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias=torch.zeros(1,dtype=torch.float64,requires_grad=True)


  def forward(self,X):
    z=torch.matmul(X,self.weights)+self.bias
    y_pred=torch.sigmoid(z)
    return y_pred

  def loss_binary_cross_entropy(self,y_pred,y):
    epsilon=1e-7 ## to avoid log(0) error
    y_pred=torch.clamp(y_pred,epsilon,1-epsilon)
    loss= -(y * torch.log(y_pred) + (1-y)*torch.log(1-y_pred)).mean()
    return loss

```

## Important Parameters

```python
learning_rate=0.1
epochs=25
```

## Training Pipeline

```python
## create a model

model=MySimpleNN(X_train)

model.weights
```

```python
model.bias
```

```python
## these all will be done in a loop which is epoch times


for epoch in range(epochs):
  ## Forward pass
  y_pred=model.forward(X_train)      ## X_train shape:455 X 30 and weights shape: 30 X 1         x_train X weights= 455 X 1

  ## loss val
  loss=model.loss_binary_cross_entropy(y_pred,y_train)
  print(f'Epoch:{epoch+1},Loss:{loss.item()}')
  ## if we try to print loss we get something like this tensor(0.8631, dtype=torch.float64, grad_fn=<NegBackward0>)
  ## What does loss.item() do? .item() extracts the actual numerical value from a tensor that contains a single element. which is float only not tensor

  ## back pass
  loss.backward()

  ## update weights and biases (parameters)
  ## in this we don't need to cal gradients
  with torch.no_grad():
    model.weights-=learning_rate*(model.weights.grad)
    model.bias-=learning_rate*(model.bias.grad)


  ## zero_gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()


```

## Evaluvation

```python
## model evaluvation


with torch.no_grad():
  y_pred=model.forward(X_test)

print(y_pred)


## we need to decide a threshold
```

```python
## model evaluvation


with torch.no_grad():
  y_pred=model.forward(X_test)
  y_pred=(y_pred>0.5).float() ## converts the boolean torch.bool to torch.float
  accuracy=(y_pred==y_test).float().mean()
  print(f"accuracy:{accuracy}")


print(y_pred)
```
