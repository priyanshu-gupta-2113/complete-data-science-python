## Plotly Dash

All the data visualization libraries were static before. If the data changes, you have to run the code again to get that effect on graph..


Using dash you can create full legit website where even if the data source changes , change will be reflected onto to the graph...

There are two components of dash:

1-Dash HTML Component: can create or design a web page ( no need to learn html and css)

2-Dash Core component: Help you built interactive dashboard

```python
import pandas as pd
import plotly.graph_objects as go
import dash_html_components as html
import dash_core_components as dcc
import dash
import dash.dependencies 
from dash.dependencies import Input,Output
```

```python
data=pd.read_csv('/Users/priyanshugupta/Desktop/plotly/gapminder.csv')
data.head()
```

```python
app=dash.Dash()
## try to create layout of your page and it will contain one user interface item 
## all html components are supported by dash library
## whatever you write inside children will be displayed in page
app.layout=html.H1(children="My first dashboard",style={'color':'red','text-align':'center'})

if __name__=='__main__':
    app.run()
```

```python
app.layout=html.Div([
    html.Div(), ## 1st contains the heading
    html.Div(), ## 2nd contains 1st graph
    html.Div() ## 3rd contains the 2nd graph
])
```



this outer rectangle will be created using Div --> Division

```python
app.layout=html.Div([
    html.Div(children=[html.H1("My First Dashboard",style={'color':'red','text-align':'center'})],style={'border':'1px white solid','float':'left','width':'100%','height':'50px'}),
    html.Div(style={'border':'1px white solid','float':'left','width':'49.5%','height':'350px'}),
    html.Div(style={'border':'1px white solid','float':'left','width':'49.5%','height':'350px'}),
])

if __name__=='__main__':
    app.run()
```

```python
app.layout=html.Div([
    html.Div(children=[html.H1("My First Dashboard",style={'color':'red','text-align':'center'})],style={'border':'1px white solid','float':'left','width':'100%','height':'50px'}),
    html.Div(children=[
        dcc.Graph(id='scatter-plot',figure={'data':[go.Scatter(x=data['pop'],y=data['gdpPercap'],mode='markers')],
        'layout':go.Layout(title='scatter plot')})

    ],style={'border':'1px black solid','float':'left','width':'49.5%'})
    
    ,
    html.Div(children=[
        dcc.Graph(id='box-plot',figure={'data':[go.Box(x=data['gdpPercap'])],
                  'layout':go.Layout(title="boxplot")}),],style={'border':'1px black solid','float':'left','width':'49.5%'}),
])

if __name__=='__main__':
    app.run()
```
