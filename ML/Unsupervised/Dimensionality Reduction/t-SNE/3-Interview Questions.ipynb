## Points of Wisdom

1. Interpreting Clusters: t-SNE can reveal clusters and local structures effectively.However, the distance between clusters or the relative position of clusters in the plot may not have a meaningful interpretation. Avoid over-interpreting global relationships

2. Axes have no meaning: Axes in t-SNE has no interpretable meaning. In PCA, there is meaning but no meaning in t-SNE. You cannot do prediction using t-SNE. You cannot apply ML algo on new features of t-SNE

3. Perplexity matters: Perplexity is a crucial hyperparameter in t-SNE. It roughly corresponds to the number of effective nearest neighbours. There is no one size that fits all values. Different values reveal different structures, so experiment with a range of values. Common values are between 5 and 50. 

4. Reproducibility: t-SNE starts with a random intialization, leading to different results each time you run it. If reproducibility is important, set a random seed. Also, multiple runs with different intializations can give a fuller picture of your data's structure. 

5. Scaling the Data: pre-processing steps like scaling or normalizing the data, especially if features are on different scales, can have significant impact on the results of t-SNE.

6. Curse of dimensionality: t-SNE can mitigate but not completely overcome the curse of dimensionality. Very high dimensional data might require other steps, like initial dimensionality reduction with PCA, before applying t-SNE

7. Learning Rate and Number of Iterations: Beyond perplexity, other parameters like the learning rate and number of iterations also impact the results. A learning rate that is too high or low can lead to poor embeddings, and insufficient iterations might mean the algorithm does not fully converge

Most Important: 

8. It is not a Silver Bullet: While t-SNE is a powerful tool, It is not a suitable tool for every kind of dataset or analysis. Sometimes other dimensionality reduction techniques like PCA, UMAP, or MDS might be more appropriate. It is one of the tool for analysis but not the only one

## Advantages:

- If done correctly, can give intutive visualizations as it preserves the local structure of the data in the lower dimensions

## Disadvantages:

- Computationally expensive
- Not very good at preserving global structure
- Sensitive to hyperparameters
- Can get stuck in local minima
- Interpretation is challenging (axes have no meaning)

# Q why use T-distribution in lower dimension ?

the Student's t-distribution has heavy tails, it gives more probability to moderately distant points than a Gaussian would. This helps alleviate the crowding problem caused by representing high-dimensional data in a low-dimensional space, allowing non-neighboring points to spread apart while preserving local neighborhoods.
