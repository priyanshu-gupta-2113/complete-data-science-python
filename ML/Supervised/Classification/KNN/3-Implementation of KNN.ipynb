```python
class KNN:
    def __init__(self,K=3,task='Classification'):
        '''
        K: no of nearest neighbours
        task: Classification or Regression
        '''
        self.K=K
        self.task=task
        self.X_train=None
        self.y_train=None

    def fit(self,X,y):
        '''Storing the training data in numpy array'''
        self.X_train=X
        self.y_train=y


    def eucledian_dist(self,point1,point2):
        ''' Computing the eucledian distance between two points'''

        distance=0.0
        for a,b in zip(point1,point2):
            distance += (a-b)**2
            return math.sqrt(distance)

    def predict_one(self,X): ## X is a test point not test set
        ''' Predicts the label or value for a single test sample'''

        distances=[]
        for i,X_train in enumerate(self.X_train):
            dist=self.eucledian_dist(X,X_train)
            distances.append((dist,self.y_train[i])) ## ex- [(1.3,'M'),(1.5,'F'),...]

        ## sort by distances
        distances.sort(key= lambda x:x[0])

        ## take K nearest neighbours
        k_nearest=distances[:self.K]
        labels=[item[1] for item in k_nearest]

        if self.task == 'Classification':
            ## We use majority voting here
            most_common_label= Counter(labels).most_common(1)[0][0]
            if type(most_common_label)==str:
                return most_common_label
            return int(most_common_label)
            

        else: ## Regression
            reg_value= sum(labels)/ len(labels)
            return float(reg_value) ## reg_value: np.float64(12.5) so we convert it to float



    def predict(self,X_test):
        ## Predicts for the whole Test set
        return [self.predict_one(x) for x in X_test]
    
```

```python
knn=KNN()
knn.fit(X_train,y_train)
y_pred=knn.predict(X_test)
y_pred
```

```python
accuracy_score(y_pred,y_test)
```
