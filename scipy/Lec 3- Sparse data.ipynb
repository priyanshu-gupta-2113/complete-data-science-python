## Sparse Data

Data that has mostly unused elements like the elements that don't carry any info.

Sparse data example - [1,0,2,0,0,3,0,0,0,0,0,0,4] 


## Dense Data

Example- [1,2,4,5,6,9,2,4] 


## How to work with sparse data



```python
## scipy has a module scipy.sparse function
## there are 2 metrics in this sparse :  CSC (Compresses sparse column) and CSR(Compressed sparse row)
```

```python
## CSR Metrics: here we will create a CSR metrics from an array 

import numpy as np
from scipy.sparse import csr_matrix

shared= np.array([0,0,0,0,0,1,1,0,2])

print(csr_matrix(shared))

## 1st coordinate represents row and 2nd coordinate represents column
## csr_matrix(shared) creates a Compressed Sparse Row (CSR) matrix.
```

```python
## Sparse matrix methods
shared=np.array([[0,0,0],[0,0,1],[1,0,2]])
print(csr_matrix(shared))
```

```python
print(csr_matrix(shared).data) ## gives you the nonzero values only 
```

```python
## to extract no of nonzero values

## what if we want to count nonzeros , we can do this via count_nonzero() method

shared=np.array([[0,0,0],[0,0,1],[1,0,2]])
print(csr_matrix(shared).count_nonzero())
```

```python
## for removing the zero elements from the matrix we will use eliminate_zeros().
## eliminate_zeros() removes explicitly stored zeros from a sparse matrix.

sharad= csr_matrix(shared)
sharad.eliminate_zeros()
print(sharad)
```

```python
## When is eliminate_zeros() useful?

## uppose you explicitly create a sparse matrix that contains stored zeros:

from scipy.sparse import csr_matrix

A = csr_matrix( ([1,0,2], ([0,1,2], [0,1,2])  ) )
print(A)
## Notice that (1,1) is stored even though its value is 0.
```

```python
A.eliminate_zeros()
print(A)
## The stored zero is removed.
```

```python
## eliminating duplicate entries with the sum_duplicates()

## sum_duplicates() is used when a sparse matrix contains multiple entries at the same coordinate.
from scipy.sparse import csr_matrix

row = [0, 0]
col = [1, 1]
data = [3, 4]

A = csr_matrix((data, (row, col)), shape=(3,3))
A.sum_duplicates()
print(A)
```

```python
## here we will convert csr to csc with tocsc() method:

shared=np.array([[0,0,0],[0,0,1],[1,0,2]])
sharadnew=csr_matrix(shared).tocsc()

## .tocsc() converts it into a Compressed Sparse Column (CSC) matrix.
print(sharadnew)
```

Matrix type

Data type
