## Python Scipy Constants

```python
## constants in scipy: scipy is more focused on scientific implementation, it provides many inbuilt
# scientifc constants and these are helpful in datascience


## printing the constant value of pi

from scipy import constants
print(constants.pi)
```

```python
## constants unit: all list of unit under the constant can be seen with dir() function

from scipy import constants
print(dir(constants))
```

```python
## Unit Categories: Metric, Binary, Mass, Angle, Time, Length, Pressure, Volume, Speed, Temp, Energy
# Temp, Power, Force



## 1. Metric (SI) Prefixes:   converts to meter

print(constants.yotta) ## 1e+24
print(constants.zetta) ## 1e+21
print(constants.exa)   ## 1e+18
print(constants.peta) ## 1e+15
print(constants.tera) ## 1e+12
print(constants.giga) ## 1e+9
print(constants.mega) ## 1e+6
print(constants.kilo) ## 1e+3
print(constants.hecto) ## 1e+2
print(constants.deka) ## 1e+1
print(constants.deci) ## 1e-1
print(constants.centi) ## 1e-2
print(constants.milli) ## 1e-3
print(constants.micro) ## 1e-6
print(constants.nano) ## 1e-9
print(constants.pico) ## 1e-12
print(constants.femto) ## 1e-15
print(constants.atto) ## 1e-18
print(constants.zepto) ## 1e-21
```

```python
## Time:
## changes to seconds 
print(constants.minute) ## 60
print(constants.hour)  ## 3600
print(constants.day) 
print(constants.week)
print(constants.year)
print(constants.Julian_year)
```
