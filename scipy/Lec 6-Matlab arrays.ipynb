Why is it called MATLAB?

Because everything in MATLAB is built around matrices and arrays.

For example:

     4 5 6]

Output:

     1     2     3
     4     5     6

## Why does SciPy mention MATLAB?

SciPy can:

Read MATLAB files (.mat) and Write MATLAB files (.mat) using scipy.io

## What is matlab array ?

In SciPy, a MATLAB array usually refers to data loaded from a MATLAB .mat file using scipy.io.

MATLAB stores variables in .mat files, and SciPy can read them.

## Exporting Data in Matlab Format

The savemat() function allows us to export data in Matlab format.

The method takes the following parameters:

filename - the file name for saving data.

mdict - a dictionary containing the data.

do_compression - a boolean value that specifies whether to compress the result or not. Default False

```python
from scipy import io


import numpy as np

arr=np.arange(10)

io.savemat("arr.mat",{"vec":arr},False)
```

## Loading a MATLAB file

```python
from scipy import io

arr=io.loadmat("arr.mat")
print(arr)

## This returns a Python dictionary containing the MATLAB variables.
```

```python
print(arr['vec'])
print(type(arr['vec']))
## So MATLAB arrays become NumPy arrays in Python.
```
