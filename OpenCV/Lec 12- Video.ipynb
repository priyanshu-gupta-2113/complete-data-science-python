## Working with videos

Video is series of images. If we show many continous images in less time, we think of it as a video

```python
import cv2
import numpy as np
```

```python
cap=cv2.VideoCapture(0) ## 0 here stands for our primary webcam and this will store the video in cap

while True:

    ret,frame=cap.read() ## ret has two values 0 and 1 and tells video is still going on or has ended --> 1 means video is still going and 0 means video has ended

    cv2.imshow('webcam',frame)  ## used to display the webcam


    if cv2.waitKey(1) & 0xFF==ord('x'):
        break


cv2.destroyAllWindows()
                        
```

```python
## to convert the video to grayscale 

cap=cv2.VideoCapture(0) 

while True:

    ret,frame=cap.read() 
    img_gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    cv2.imshow('webcam',img_gray)


    if cv2.waitKey(1) & 0xFF==ord('x'):
        break


cv2.destroyAllWindows()
                        
```

## to store the video stream 


```python
import cv2

cap = cv2.VideoCapture(0)

ret, frame = cap.read()
h, w = frame.shape[:2]

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('output.mp4', fourcc, 20.0, (w, h))

print("Writer opened:", out.isOpened())

while True:
    ret, frame = cap.read()

    if not ret:
        break

    out.write(frame)

    cv2.imshow("webcam", frame)

    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

```python

## Gray version of video 

import cv2

cap = cv2.VideoCapture(0)

ret, frame = cap.read()


h, w = frame.shape[:2]

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('output2.mp4', fourcc, 20.0, (w, h),isColor=False) ## for grayscale

while True:
    ret, frame = cap.read()
    img_gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    out.write(img_gray)
    cv2.imshow("webcam", img_gray)
    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```
