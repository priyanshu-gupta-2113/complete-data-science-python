## Black and White Vs GrayScale Image


## Definition
Black-and-white (binary, bilevel): each pixel has only two possible values — black or white. Technically a 1-bit image (0 or 1).

Grayscale: each pixel can represent many shades between black and white. Common formats are 8-bit (256 levels), 16-bit, or floating-point, giving smooth tonal variation.

## Visual and perceptual differences

Black-and-white: high-contrast, posterized look with hard edges and no intermediate tones. Fine gradations (shadows and highlights) are lost unless dithered.

Grayscale: continuous tonal transitions, preserves gradients, subtle shadows, midtones and texture.

## File formats and storage

Black-and-white: stored as 1-bit per pixel or as indexed color with two palette entries; very small file size.

Grayscale: stored as 8-bit (0–255) or higher per pixel; larger file size, more detail. Common formats: PNG, TIFF, JPEG (luminance channel), BMP.

## Use cases

Black-and-white: line art, scanned text, fax, simple logos, thresholded images, some binary masks and segmentation maps.

Grayscale: photography, medical imaging (X‑ray), image analysis where intensity variation matters, artistic monochrome photos.
