```python
import cv2
import numpy as np
```

```python
img=cv2.imread('/Users/priyanshugupta/Desktop/OpenCV/fruits.png')
## synatx: cv2.cvtColor(img to convert, cv2.COLOR_BGR2GRAY)                  cvtColor= convert color
''' 
different conversions are there
cv2.COLOR_BGR2GRAY
cv2.COLOR_BGR2RGB
cv2.COLOR_BGR2LUV
'''

## In opencv it is blue green red (BGR) not red green blue(RGB)

print(img.shape)
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
print(img_gray.shape)
```

```python
cv2.imshow("window", img_gray)

cv2.waitKey(2000)  # Display for 2000 ms = 2 seconds
cv2.destroyAllWindows()


## cv2.waitKey(0) → waits indefinitely until a key is pressed.
## cv2.waitKey(2000) → waits 2 seconds and then continues automatically.
## cv2.destroyAllWindows() → closes the image window after the wait.
```
