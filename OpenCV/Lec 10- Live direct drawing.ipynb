```python
## on dragging we want to draw rectangles on image


## Used in Object Detection
import numpy as np
import cv2
img=np.zeros((512,512,3))
im=cv2.rectangle(img,pt1=(100,100),pt2=(300,300),color=(255,0,0),thickness=3)
im2=cv2.rectangle(im,pt1=(100,100),pt2=(300,300),color=(255,0,0),thickness=-1)
im2=cv2.circle(im2,center=(100,100),radius=50,color=(0,0,255),thickness=3)
im2=cv2.line(im2,pt1=(0,0),pt2=(512,512),thickness=2,color=(0,255,0))
im2=cv2.putText(im2,org=(0,400),fontScale=4,color=(0,255,255),thickness=2,fontFace=cv2.FONT_ITALIC,text="Hello")
```

```python
flag=False
ix=-1
iy=-1
def draw(event,x,y,flags,params): ## we draw after dragging and not just after clicking 
    global flag,ix,iy
    if event==1:
        flag=True  ## i can draw if i want 
        ix=x
        iy=y


    elif event==0:
        if flag==True:
            cv2.rectangle(img,pt1=(ix,iy),pt2=(x,y),color=(0,255,255),thickness=-1)
    
    elif event==4:
        flag=False
        cv2.rectangle(img,pt1=(ix,iy),pt2=(x,y),color=(0,255,255),thickness=-1)


cv2.namedWindow(winname='window')
cv2.setMouseCallback('window',draw)

while True: ## infinite loop
    cv2.imshow('window',im2)
    ## condition for closing infinite loop
    if cv2.waitKey(1) & 0xFF ==ord('x'):  ## apne keyboard se x wala button pres kiya
        break
```
