## Drawing shapes

```python
## Used in Object Detection
import numpy as np
import cv2
img=np.zeros((512,512,3))
```

```python
cv2.imshow('window',img)
cv2.waitKey(19)
```

```python
## Rectangle
## for rectangle we need two points top most left corner and bottom most right corner
im=cv2.rectangle(img,pt1=(100,100),pt2=(300,300),color=(255,0,0),thickness=3)
cv2.imshow('window',im)
cv2.waitKey(19)
```

```python

## to fill the rectangle with same color use thickenss=-1
im2=cv2.rectangle(im,pt1=(100,100),pt2=(300,300),color=(255,0,0),thickness=-1)
cv2.imshow('window',im2)
cv2.waitKey(19)
```

```python
## Circle
## give center of circle and radius 
im2=cv2.circle(im2,center=(100,100),radius=50,color=(0,0,255),thickness=3)
cv2.imshow('window',im2)
cv2.waitKey(19)
```

```python
## Line

im2=cv2.line(im2,pt1=(0,0),pt2=(512,512),thickness=2,color=(0,255,0))
cv2.imshow('window',im2)
cv2.waitKey(19)
```

```python
## Text
## fontscale tells size of color
im2=cv2.putText(im2,org=(0,400),fontScale=4,color=(0,255,255),thickness=2,fontFace=cv2.FONT_ITALIC,text="Hello")
cv2.imshow('window',im2)
cv2.waitKey(19)
```
