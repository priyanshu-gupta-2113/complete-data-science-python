## Flipping an image

```python
import cv2
img=cv2.imread('/Users/priyanshugupta/Desktop/OpenCV/fruits.png')
img.shape
```

```python
## flip codes: 0 ,1 and -1

## 0 -> vertical flipping
## 1 -> horizontal flipping
## -1-> combination of horizontal and vertical flipping
img_flip=cv2.flip(img,0)
cv2.imshow('window',img_flip)
cv2.waitKey(19)
```

```python
img_flip=cv2.flip(img,1)
cv2.imshow('window',img_flip)
cv2.waitKey(19)
```

```python
img_flip=cv2.flip(img,-1)
cv2.imshow('window',img_flip)
cv2.waitKey(19)
```
