## Context Window

Y = f(x) -> where x is input tokens. But is there any max limit of tokens you can send to LLM ? Yes

Context Window is the amount of text an LLM can read and remember at one time before answering. 

Think of an LLM like a camera.

- LLM = camera
- Context window = camera lens
- Text/messages = scene in front of the camera

If the lens is small, the camera can capture only a small portion of the scene.

Similarly, if the context window is small, the LLM can process only a limited amount of text at one time.

If the context window is large, the LLM can process much more text.

Therefore, Context window = maximum amount of information the model can consider in a single request.


```python
# Example

# Suppose an LLM has a context window of 100,000 tokens.

# You can potentially send:

# Previous conversation
# +
# System instructions
# +
# Your documents
# +
# Current question

# as long as the total stays within the model's context limit.
```

## Parametric Knowledge

During training, an LLM sees enormous amounts of data.

Conceptually: Training data -> Patterns learned -> Model parameters

The information learned during training is encoded into the model's parameters.

That's why it's called:

Parametric knowledge

When you train an LLM on huge amount of data (whole internet), the LLM gathers knowledge from the training data and finds patterns in it, the whole knowledge learnt is stored in parameters. That is why it is known as parameteric knowledge. 

When a model gets trained, whenever at inference time, you ask a question to model then the parametric knowledge is used to answer that question

For example, the model may have learned:

1. Paris → capital of France
2. 2 + 2 → 4
3. Python → programming language

This knowledge is available because it was learned during training.
the model hasn't been trained on your private conversation with it.

So if you tell it:

My name is Priyanshu.

That information isn't automatically written into its model parameters.


## Then how can it know your name?

This is where in-context learning comes in.

Suppose we send a Prompt: 

My name is Priyanshu.

What is my name?


The model doesn't need to have learned your name during training.

It can read the information inside the current input and use it to answer.

That's the basic idea behind:

In-context learning

## In-context Learning:

It is an emergent ability that allows an LLM to use information and patterns present in the prompt itself, in addition to its trained parameteric knowledege, to generate an answer.

The LLM can use information and patterns present in the current prompt/context to generate its answer, without modifying its parameters.

Imagine you give the model this:

My company's internal document says:

Production increased by 17% last quarter.

Question:
What was the production increase?

The model may answer:

17%

It doesn't mean the model has permanently learned your company's document.

It simply used the information that you supplied inside the current context.

## How short term memory is implemented in Chatbots ?

Chatbots have a concept of Conversation. Conversation is one session with user and chatbot. 

Suppose you open ChatGPT and have this conversation:

- You: What is machine learning?
- ChatGPT: Machine learning is...
- You: Explain supervised learning.
- ChatGPT: Supervised learning is...
- You: Give me an example.
- ChatGPT: ...

This entire sequence is one conversation.

You can think of it as:


```python

# Conversation 1
# │
# ├── User: What is ML?
# ├── AI: ...
# ├── User: What is supervised learning?
# ├── AI: ...
# └── User: Give me an example

# Then tomorrow you start a completely different chat:

# Conversation 2
# │
# ├── User: What is Python?
# ├── AI: ...
# └── User: Explain lists

# These are two separate conversations.
```

## Q: What is Short-Term Memory?

The chatbot needs to remember what you said earlier in the same conversation.

For example:

- User: My name is Priyanshu.
- AI: Nice to meet you, Priyanshu.

- User: What is my name?
- AI: Your name is Priyanshu.

How did the AI know? Because the previous messages are available as conversation history.

Conceptually, the application might maintain something like:

messages = [ 
    
    {"role": "user", "content": "My name is Priyanshu"},

    {"role": "assistant", "content": "Nice to meet you, Priyanshu"},

    {"role": "user", "content": "What is my name?"}
    
    ]

When the next question comes, the application sends the relevant conversation history to the LLM.

So:

Conversation -> Message history -> LLM -> Response

That message history acts as the chatbot's short-term memory.

## Q: Why is it called "short-term"?

Because this memory belongs only to the current conversation.

Imagine: Conversation A

- User: My project is about fraud detection.
- User: I am using Random Forest.

The chatbot knows:

Project → Fraud Detection

Model → Random Forest

Now you start Conversation B:

User: What model should I use for image classification?

The chatbot should not automatically assume:

- Project = fraud detection
- Model = Random Forest

because those belonged to Conversation A.

Conceptually:

Conversation A -> Short-term memory A

Conversation B -> Short-term memory B

They are separate.


## Conversation-scoped memory

Short-term memory is conversation-scoped.

```python
#               Chatbot
#                  │
#        ┌─────────┴─────────┐
#        ↓                   ↓
#  Conversation A       Conversation B
#        │                   │
#        ↓                   ↓
#  Memory A             Memory B

# Memory A belongs to A.

# Memory B belongs to B.

# They don't automatically mix
```

## Conversation = Thread

In many AI systems, conversation is called a thread.

So:

Conversation ≈ Thread

For Example:

```python
# Thread 1
# │
# ├── Message 1
# ├── Message 2
# ├── Message 3
# └── Message 4


# Thread 2
# │
# ├── Message 1
# ├── Message 2
# └── Message 3

# Therefore when you read: "Short-term memory is thread-scoped"

# it basically means: The chatbot's short-term memory belongs to one particular conversation/thread.
```

Think of it this way:

Thread: Thread ID = 123

contains:

- Message 1
- Message 2
- Message 3
- Message 4

That's the chatbot's short-term conversational context.

Another thread: Thread ID = 456 contains completely different messages.

```python
# Thread 123                 Thread 456
# ──────────                 ──────────
# Memory A                   Memory B
#     │                          │
#     ↓                          ↓
# Conversation A             Conversation B
```

```python
## How this looks in actual chatbot

conversations = {
    "thread_123": [],
    "thread_456": []
}

conversations["thread_123"].append(
    {"role": "user", "content": "My name is Priyanshu"}
)


## suppose the AI responds

conversations["thread_123"].append(
    {"role": "assistant", "content": "Nice to meet you!"}
)

## Later you load next conversation
conversations["thread_123"]

## while conversations["thread_456"] contains history of another conversation
```

## Short-term memory does not necessarily mean the entire conversation is literally stored in one Python variable forever.

One possible way: Database -> Retrieve messages for thread_id -> Possibly summarize / trim old messages -> Build context -> LLM

So the key architectural idea is scoping the conversational context to a thread, not necessarily the specific implementation using a Python messages list.

```python
#                     CHATBOT
# #                        │
# #           ┌────────────┴────────────┐
# #           │                         │
# #        Thread 1                  Thread 2
# #      (Conversation 1)          (Conversation 2)
# #           │                         │
# #           ↓                         ↓
# #      Short-term                Short-term
# #        memory                    memory
# #           │                         │
# #      ┌────┴────┐               ┌────┴────┐
# #      │         │               │         │
# #    User      AI              User       AI
# #    msg       msg             msg        msg
# #      │         │               │         │
# #      └────┬────┘               └────┬────┘
# #           │                         │
# #           └── Context A             └── Context B

# Conversation = Thread (in many chatbot frameworks).
# Short-term memory = the message/context history of that thread.
# Different threads have separate short-term memories
```

## Problems with short term Memory:

Short-term memory remembers the current conversation, but it does not automatically give the AI persistent memory across conversations.

## Problem 1 - Short term memory is fragile



Suppose your chatbot stores conversation history in:

messages = []

As the conversation continues:

messages:
- Message 1
- Message 2
- Message 3
- Message 4
- Message 5
- ...

Now imagine your application/server crashe or your Python process restarts or your application gets redeployed. The variable messages is gone. Therefore:

Server crashes -> messages variable disappears ->  Conversation history disappears -> AI loses the context



***Example of the problem***

Imagine you're talking to your chatbot:

Conversation 1

- You: My project is about fraud detection.
- AI: Okay.
- You: I'm using Random Forest.
- AI: Great.
- You: Explain how to improve it.
- AI: ...

All of this is currently inside: messages

Now you click New Chat. The application starts a new conversation.

Conversation 2

messages = []

You start talking:

- You: Explain CNN.
- AI: ...

Now you suddenly go back to Conversation 1. If the application had only stored the messages in RAM, it may have nothing to reload.

So:

Conversation 1 ->  messages variable -> New Chat -> messages reset -> Conversation 1 history lost


# Solution: Persistence + Database

The solution is: Don't rely only on the messages variable. Persist the conversation somewhere.

Chatbot -> Short term memory -> Database

create a thread id, for example thread_id = 1

Then store:

```python
# thread_id	     role	         message
# 1	             user	        My project is fraud detection
# 1	           assistant	        Okay
# 1	             user	       I'm using Random Forest
# 1	           assistant	       Great
```

Now start another conversation: thread_id = 2

Store its messages separately

```python
# thread_id	  role	      message
# 2	          user	    Explain CNN
# 2	        assistant	  CNN is...
```

```python
# Now if you return to thread 1:

# thread_id = 1
#        ↓
# Database
#        ↓
# Retrieve messages
#        ↓
# Load into messages
#        ↓
#       LLM

# The conversation continues exactly where you left off. This is what we mean by persistance
```

RAM / messages variable -> Temporary memory

Database -> Persistent storage

So STM itself is still thread-scoped, but the conversation history can be persisted so that you can reconstruct that STM when you reopen the thread.

A model can process only a limited number of tokens in one request.

For example, conceptually:

Context window = 100,000 tokens

Your conversation starts:

- Message 1
- Message 2
- Message 3
- ...
- Message 50

No problem. But imagine the conversation continues for weeks:

- Message 1
- Message 2
- ...
- Message 500
- ...
- Message 1000
- ...
- Message 2000

If your application sends the entire conversation history every time, the input becomes enormous. We describes exactly this accumulation problem.

Suppose the model can handle 100,000 tokens and we give it 150,000 tokens, 150,000 > 100,000 , The context exceeds the model's context window.

This can cause the model to fail to properly process the conversation and potentially produce incoherent responses or hallucinations

In Trimming, you keep the most recent messages (Ex: keep last 50 messages)

# Problem 3.A: Learning does not compound 

This is slightly different from remembering a single preference. Imagine you're teaching the AI how you want SQL queries written

Conversation 1:

You: Give me top 5 customers.

AI: SELECT ... subquery ...

You: Use a window function instead.

AI: SELECT ... ROW_NUMBER() OVER (...)

So you've effectively taught the AI something about your preferred approach.

But then you start Conversation 2.

You ask: Give me top 5 products.

The AI may again give you a subquery. Why? Because the previous conversation's learning isn't automatically available. We describe this as the user's effort not compounding over time.


```python
# Think:

# Conversation 1
#    ↓
# You teach AI something
#    ↓
# Conversation ends
#    ↓
# Learning stays there

# Conversation 2
#    ↓
# Start again
#    ↓
# Teach AI again

# Instead, what you want is:

# Conversation 1
#    ↓
# Learn preference A
#    ↓
# Conversation 2
#    ↓
# Learn preference B
#    ↓
# Conversation 3
#    ↓
# Learn preference C
#    ↓
# Better user understanding

# That's compounding memory.
```

## Problem 3.B - Cross thread reasoning

Suppose yesterday you had:

Conversation 1

You: My model is giving poor accuracy.

AI: Try feature engineering.

Today you start Conversation 2

You: What solution did we discuss yesterday for my model?

With only thread-scoped STM, the AI cannot automatically access Conversation 1.

Therefore:





```python
# Thread 1
#    ↓
# Yesterday's discussion
#        X
#        │
#        │ no automatic access
#        ↓
# Thread 2
```

So cross-thread questions such as:

"What did we discuss yesterday?"

or: "What solution did we find last time?" cannot be answered from STM alone. We call cross-thread reasoning and explains why thread-scoped STM prevents it.

## Problem 3.c: Cannot learn about user preferences

We need a completely new type of memory. And this is where the lecture is transitioning from Short-Term Memory → Long-Term Memory.

Therefore the AI can't naturally evolve into a deeply personalized assistant using STM alone.

## Solution for all problems: We need a new kind of memory

Short term memory could hold a single conversation data inside. As soon as it is removed, it will get destroyed. But Long term is a memory that exists outside of a conversation and exists for a long period of time

## Long term Memory

Episodic Memory: Past events

Semantic Memory: user preferences and facts

Procedural Memory: Strategies, rules and learnings

## How Long-Term Memory Works (High-Level Architecture)



Short-term memory remembers the current conversation. Long-term memory extracts important information from conversations, saves it permanently, retrieves only what is relevant later, and puts that information back into the current context for the LLM.

LTM into 4 Steps:

1. Creation / Update
2. Storage
3. Retrieval
4. Injection

```python
    #              USER
    #                │
    #                ↓
    #          Conversation
    #                │
    #                ↓
    #       ┌─────────────────┐
    #       │ 1. CREATION     │
    #       │ Find memories   │
    #       │ worth saving     │
    #       └────────┬────────┘
    #                ↓
    #       ┌─────────────────┐
    #       │ 2. STORAGE      │
    #       │ Save memories   │
    #       │ permanently     │
    #       └────────┬────────┘
    #                ↓
    #           MEMORY STORE
    #                │
    #                │
    #          New conversation
    #                │
    #                ↓
    #       ┌─────────────────┐
    #       │ 3. RETRIEVAL    │
    #       │ Find relevant   │
    #       │ memories        │
    #       └────────┬────────┘
    #                ↓
    #       ┌─────────────────┐
    #       │ 4. INJECTION    │
    #       │ Put memories    │
    #       │ into STM/context│
    #       └────────┬────────┘
    #                ↓
    #               LLM
    #                ↓
    #             Response
```

## Step 1- Creation/Updation

Find out whether anything in the current conversation is worth remembering beyond this conversation. This is sometimes called memory creation or memory update.

Is there anything that has just happened that is worth remembering beyond this conversation?

Example

Suppose the user says:

"I prefer Python over Java."

The system identifies:

"I prefer Python over Java"

as a memory candidate.

Why?

Because this information could be useful in future conversations.

Compare that with:

"Okay, thanks."

That's probably not worth saving.

***System looks at:***

1. User Messages
2. Model Responses
3. Tool Outcomes

***What happens in this step:***

The system:

1. Extracts memory candidate
2. filters out noise
3. decides scope (user/agent/app)
4. decides whether to :
- Create a new memory
- Update an existing one
- ignore it

User-level

Something specifically about the user.

Example:

User prefers Python.

This could potentially be useful across the user's conversations.

App-level

Something relevant to the application.

For example, a chatbot application might have some persistent information relevant to the application itself.

Agent-level

Something relevant to a particular AI agent.

For example:

Agent should use a particular procedure.

The key idea is that you must decide who/what this memory belongs to before storing it.

## Step 2: Storage

Now suppose the system has decided: "Yes, this information should become long-term memory.". It answers where and how do we keep this memory ?.

You need to store it somewhere durable.

Why durable ?

Because you want the memory to survive:

- restarting the application,
- server crashes,
- ending the conversation,
- future conversations.

# What storage actually means

1. Writing memory to a durable store
2. Assigning identifiers and metadata
3. Making it survive restarts and crashes

Storage also adds metadata

The system doesn't necessarily just save:

"I prefer Python."

It can also attach:

Memory -> Content, identifier, scope, metadata

For example, conceptually:

{
  "memory": "User prefers Python over Java", "scope": "user", "id": "memory_123"
}

## What can you use as a memory store ?

1. A relational database
2. a Key-Value store
3. a log/text file
4. a vector database (for semantic lookup)

different storage approaches for:

- episodic memory
- procedural memory
- semantic memory

The important point is:

There isn't one mandatory database technology for all long-term memory.

The storage mechanism depends on the memory and retrieval requirements.

## Retrieval:

Given the current situation, what should I remember right now ?

A new conversation begins.

The user asks:

"Give me code for this algorithm."

The system doesn't take every memory in the database and send it to the LLM. Instead, it asks: Do I need memory to answer this question?

# What retrieval involves:

The system:

1. looks at current input
2. decides what memory is needed
3. searches memory stores
4. selects a small,relevant subset


Key Point: Retrieval is selective, not exhaustive

```python
# 1000 memories in database
#           ↓
#       Search
#           ↓
#      5 relevant memories
#           ↓
#          LLM


# NOT:

# 1000 memories
#       ↓
#      LLM

# Why?

# Because most memories are irrelevant to the current question.
```

# Example of selective retrieval

Suppose your memory database contains:

Memory 1 → User prefers Python

Memory 2 → User likes concise explanations

Memory 3 → User works on ML

Memory 4 → User likes travelling

Memory 5 → User prefers dark mode

Memory 6 → User's favorite food

...

Memory 500 → ...

Now user asks:

"Explain Random Forest in simple terms."

Relevant memories might be: User works on ML and User prefers simple explanations.

So retrieval gives:



```python
# Memory Store
#      │
#      ├── Memory 1 ─────── relevant ✅
#      ├── Memory 2 ─────── relevant ✅
#      ├── Memory 3 ─────── relevant ✅
#      ├── Memory 4 ─────── irrelevant ❌
#      ├── Memory 5 ─────── irrelevant ❌
#      └── ...
```

## Injection:

Long-term memory does not directly interact with the LLM. You search Long term memory and then pull relevant memories into short term memory and this becomes part of your context window and this go to LLM

Long term memory first becomes part of conversation buffer in short term memory

Answers how does memory actually influence the model ?

# What injection really is ?

- Retrieved memory is inserted into STM
- it becomes part of the prompt
- the model sees it as just more tokens
