planning and executing a trip from Delhi to Goa.

The core idea

The instructor is comparing two experiences:

- Traditional approach: The human does everything manually.
- AI Agent approach: You give the agent a goal, and it plans, makes decisions, uses tools, asks for your preferences, and executes tasks for you.

## 1. Traditional way of planning a trip

Suppose you want to travel:

Delhi → Goa, May 1 → May 7

Traditionally, you would have to do multiple tasks yourself.

Step 1: Book transportation

You need to:

1. Go to a train or flight booking website.
2. Enter source and destination.
3. Enter travel dates.
4. Search for available options.
5. Compare prices.
6. Check seat availability.
7. Select your preferred class.
8. Make payment.

Then you have to repeat the process for the return journey.

Step 2: Book accommodation

Next, you need to:

1. Search for hotels.
2. Enter check-in and check-out dates.
3. Enter the number of people.
4. Compare hotels.
5. Check reviews.
6. Compare prices.
7. Check facilities such as breakfast.
8. Select and pay.


Step 3: Plan activities

Then you need to research:

1. Which beaches to visit.
2. Which attractions to visit.
3. Which forts or churches to see.
4. Where to party.
5. How to travel locally.
6. Whether tickets are required.

Finally, you have to create an itinerary for all seven days.

main point is that this entire process requires research, decision-making, comparing information from multiple sources, communication, and payments, making it time-consuming and complicated.

## What problem does an AI Agent solve?

Today's websites require humans to navigate multiple interfaces and manually perform multiple steps.

```python
# User
#  ↓
# Search website
#  ↓
# Fill form
#  ↓
# Compare options
#  ↓
# Choose
#  ↓
# Go to another website
#  ↓
# Repeat
#  ↓
# Make decisions
#  ↓
# Book

# This can be difficult, especially for people who are not comfortable using technology.
```

```python
# # An AI Agent provides a more natural interface:

# User says what they want
#         ↓
# AI Agent understands the goal
#         ↓
# AI Agent performs research
#         ↓
# AI Agent uses tools/websites/APIs
#         ↓
# AI Agent compares options
#         ↓
# AI Agent asks for approval
#         ↓
# AI Agent takes actions

# So instead of learning how to use every website, you simply tell the AI what you want to achieve.
```

## Giving the AI Agent a Goal

The user might simply say:

"Can you create a budget travel itinerary from Delhi to Goa from May 1 to May 7?"

This is called the user goal.

The agent then extracts important information:



```python

# | Information | Extracted by Agent |
# | ----------- | ------------------ |
# | Source      | Delhi              |
# | Destination | Goa                |
# | Dates       | May 1–May 7        |
# | Duration    | 7 days             |
# | Preference  | Budget travel      |


# Then the agent creates an internal goal, something like:

# Plan a complete itinerary while optimizing cost.

# This is an important concept.
```

## User Goal vs Agent's Internal Plan

User Goal = "Plan my budget Goa trip." -> Agent understands the goal -> Internal Objective = "Find affordable transport + stay + local travel + activities."

The user doesn't need to specify every tiny step. The agent breaks the large objective into smaller tasks

## Planning: Breaking the Goal into Tasks

The AI Agent might create a plan like:

Goal: Plan a budget Goa trip

- Task 1 → Find transportation
- Task 2 → Find accommodation
- Task 3 → Arrange local transportation
- Task 4 → Plan activities
- Task 5 → Arrange return travel
- Task 6 → Calculate total cost
- Task 7 → Book everything

This ability to decompose a complex goal into smaller tasks is one of the important characteristics of an AI Agent.

## Tool Use

The agent cannot magically know real-time information.

For example, to know:

- Available trains
- Available flights
- Hotel prices
- Seat availability

It needs access to tools or APIs.


```python

# AI Agent
#    │
#    ├── Train API
#    │
#    ├── Flight API
#    │
#    ├── Hotel API
#    │
#    ├── Payment API
#    │
#    └── Calendar API

# The AI Agent decides which tool to use depending on the current task.
```

Decision Making

Notice something important here. The AI Agent doesn't simply return raw data.

A normal search system might say: Here are 50 trains.

But an AI Agent can potentially do:

I found 50 trains.

Let me compare them.

Criteria:
- ✓ Price
- ✓ Duration
- ✓ Availability
- ✓ User preference

Best option → Train X

So an Agent performs a loop:

OBSERVE → THINK → DECIDE → ACT

Example:

Observe: Available train and flight options.

Think: User wants budget travel.

Decide: Train is better than flight.

Act: Recommend the train.

## The Agent asks for Human Approval

The agent does not necessarily make every decision independently.

For example: "I found a hostel for ₹650/night. Should I book it?"

The user says:

"Yes."

Then the agent moves to the next step.

This creates a pattern called Human-in-the-Loop.

Agent researches -> Agent recommends -> Human approves/rejects -> Agent continues

## What is AI Agent ?

Definition: An AI Agent is an intelligent system that receives a high-level goal from a user, and autonomously plans, decides and executes a sequence of actions by using external tools,APIs or knowledge sources. It maintains the context, reasoning over multiple steps, adapting to new information and optimizing for the intended outcome

Agent = LLM + External Tools + Memory

- It uses LLM as reasoning mechanism 
- It uses tools for execution 

## Characteristics of AI Agent:

1. Goal Driven : You tell the agent what you want, not how you do it

2. Autonomous planning: Agent breaks down the problem and sequences tasks on its own (Breaks big tasks into small tasks)

3. Tool Using: Agent calls APIs, calculators, search tools etc

4. Context aware: Maintains memory across steps to inform future actions

5. Adaptive: Rethinks plan when things change (eg API Fails, no data)


## ReAct

Re -> Reasoning and Act -> Acting

ReAct is a design pattern used in AI agents that stands for Reasoning + Acting. It allows a LLM to interleave internal reasoning (Thoughts) with external actions (like tool use) in a structured and multi-step process.

Instead of generating an answer in one go, the model thinks step by step, deciding what it needs to do next and optionally calling tools (APIs, calculators, web searches etc) to help it.

Before using agents, what we used to do:

query -> LLM -> Response (This is a single turn interaction)

But in ReAct, we solve the problem in multi step process in a structured manner.


Three things happen in ReAct:

1. Thought
2. Action
3. Observation

We perform these 3 things in a loop till we find final answer

Example: Let's say we have a ReAct agent and we ask this agent, can you tell me population of capital of France ?





There are 3 loops of (Thought, Action and observation)

In the end, loop breaks

Note: Whatever the agent thinks and performs action is visible to user, the thought trace is visible to user. We can see what the agent is thinking and what it is doing

## ReAct is useful for:

1. Multi-step process
2. Tool-augmented tasks (web search, database, lookup etc)
3. Making the agent's reasoning transparent and auditable


It was first introduced in the paper:
"ReAct: Synergizing Reasoning and Acting in Language Models"

## Agent and Agent Executor

We will see how agent and agent executor implements react design patterns.

we need to constantly run the loop of thought,action,observation in ReAct untill final ans is achieved. AgentExecutor orchestrates the entire loop:

1. Sends inputs and previous messages to the agent
2. Gets the next action from the agent 
3. Executes that tool with provided input
4. Adds the tool's observations back into history
5. Loops again with updated history until the agent says Final Answer



STEP 1 :
Sends two things to agent:

1. User query
2. Thought trace . (thought,action,observation sequence )

STEP 2: thinks, takes action based on the input. 


action:search_tool
action input: Capital of France


STEP 3:

Search tool gives result back to the agent executor 

gets paris

now agent executor adds Paris in thought trace


STEP 4: It again sends user query + Updated thought trace

again this happens

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
import requests
from dotenv import load_dotenv
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_classic.agents import AgentExecutor,create_react_agent
from langchain_classic import hub ## prompts are stored in langchain hub and we can directly download prompt
load_dotenv()
```

```python
search_tool=DuckDuckGoSearchRun()

results=search_tool.invoke('Top news in India today ?')
results
```

```python
model=ChatGoogleGenerativeAI(model='gemini-3.6-flash')
```

```python
## Step 2: Pull the ReAct Prompt

from langsmith import Client 
client = Client()

prompt = client.pull_prompt(
    "hwchase17/react",
    dangerously_pull_public_prompt=True
)


## For every AI Agent, there is a design pattern on the back. The design patter we are going to use is called ReAct
```

```python
print(prompt.template)
```

agent_scratchpad is memory/ thought trace --- Used to maintain memory

```python
## Step 3: Create the ReAct Agent manually with inserted prompt

agent=create_react_agent(
    llm=model,
    prompt=prompt,
    tools=[search_tool]
)
```

Agent is the main guy, who breaks down the problem and decide the tool. Agent Executor is the guy who agrees with the agent and executes the step


AgentExecutor uses the tool and gives the result back to the agent

```python
# Step 4: Wrap it up with Agent Executor 
## Create AgentExecutor object

agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tool],
    verbose=True, ## to show the thought trace
    handle_parsing_errors=True
)
```

```python
## Step 5: Invoke

response= agent_executor.invoke({'input':'3 ways to reach goa from delhi'})
```

```python
## Updated code

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.agents import create_react_agent,AgentExecutor
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.tools import tool
from dotenv import load_dotenv
load_dotenv()
```

```python
@tool
def get_weather_data(city:str)-> str:
    """
    This function fetches the current weather data for a given city
    """
    url=f'https://api.weatherstack.com/current?access_key=4d1d8ae207a8c845a52df8a67bf3623e&query={city}'
    response=requests.get(url)
    return response.json()

@tool
def search(query: str) -> str:
    """Search the web for information. Keep queries short and concise."""
    
    query = query[:200]   # Prevent huge queries
    
    return search_tool.invoke(query)
```

```python
model=ChatGoogleGenerativeAI(model='gemini-3.6-flash')
```

```python
from langsmith import Client 
client = Client()

prompt = client.pull_prompt(
    "hwchase17/react",
    dangerously_pull_public_prompt=True
)

agent= create_react_agent(
    tools=[search,get_weather_data],
    llm=model,
    prompt=prompt
)

agent_executor=AgentExecutor(
    agent=agent,
    verbose=True,
    tools=[search,get_weather_data],
    handle_parsing_errors=True ## This will send the parsing error back to Gemini and give it another chance. However, this is not the ideal fix. Gemini may continue producing the entire trajectory.
)
from langchain_core.prompts import PromptTemplate

template = prompt.template + """

IMPORTANT RULES:
- Perform only ONE action at a time.
- After producing an Action and Action Input, STOP immediately.
- Do NOT generate an Observation yourself.
- The system will provide the Observation after the tool is executed.
- Do not generate multiple Actions in a single response.
- Only generate Final Answer when no more tools are required.
"""

prompt = PromptTemplate(
    template=template,
    input_variables=prompt.input_variables
)

response=agent_executor.invoke({'input':'Find capital of Madhya Pradesh then find its current weather condition'})
print(response)
```
