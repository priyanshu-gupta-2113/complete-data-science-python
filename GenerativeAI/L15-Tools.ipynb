## Tools in Langchain

Tools ---> Tool Calling --> Agents

Think of it as increasing levels of capability.


# Level 1- Tools

You give an LLM access to external functions.

Ex- def get_weather(city):

    ...

The function is the tool for the LLM.

# Leval 2- Tool Calling

The LLM can decide: "I need weather information, so I should call get_weather." The LLM doesn't necessarily execute the python function itself. It produces a ***structured tool-call request***, and the application/framework executes the function.


## Level 3 — Agents

Now the LLM can repeatedly:

Think -> Choose a Tool -> Call Tool -> Observe the result -> Thinks again -> Choose another tool -> Call Tool -> .... -> Final Answer

## What is a Tool ?

Tool is just a python function (or an API) that is packaged in a way that LLM can understand and call it when needed

```python
def add(a,b):
    return a+b
```

A human programmer knows:

function name = add

input = a, b

output = number

But an LLM doesn't automatically know that this function exists.

You need to expose it to the LLM with information such as:

Tool name: add

Description: Adds two numbers

Arguments: a-> number, b-> number

Now the LLM can understand if the user asks me to add two number, I can use this tool.

## Limitations of LLM:

LLM itself cannot execute any tasks apart from thinking and speaking. It can't do the following tasks:

- Access live data (weather,news)

- Do reliable Mathematical Computation: LLM can perform mathematical reasoning but they are not inherently calculators.  Ex: Calculate 827364 X 928374. An LLM might reason about it, but a calculator tool can produce the exact result much more reliably

- Calling APIs: Suppose you have STOCK API ... eg def get_stock_price(symbol):    to call and execute the api you need a application to excute the function.

- Running Code: Suppose a user asks: 'Analyze this CSV and calculate the average revenue.' The LLM can write Python code, but simply generating: df['revenue'].mean() doesn't execute it. You need a code execution tool. Ex: def execute_python(code): ...

- Interacting with databases: Suppose your company has: PostgreSQL with tables like customers, orders, products. The LLM doesn't automatically knows what is inside the database. You could provide a database tool. example def reun_sql(query):... The LLM can generate SELECT AVG(revenue) FROM Order;

## Custom Tool

A custom tool is a tool that you define it by yourself

Use them when:

- You want to call your own APIs 
- You want to encapsulate business logic
- You want LLM to interact with your database,product or app

In nutshell, if you already have an application and you want to create an agent for that application itself, you create custom tools for your agent so that agent can interact with infrastructure of your app (interact with database of your app)

```python
from langchain_core.tools import tool
```

```python
## Step 1- Create a function 


def Multiply(a,b):
    ''' Multiply two numbers''' ## This docstring helps LLM to understand what does this tool actually perform ?
    return a*b



## Step 2- add type hints

def multiply(a:int, b:int)-> int:
    ''' Multiply two numbers'''
    return a*b                   ## This helps LLM to understand what kind of data it can give as an input and what output value will it receive 


## Step 3- add Tool decorator
               
@tool                            ## converts this function to a special function. LLM now can communicate with this special function 
def multiply(a:int,b:int)->int:
    ''' Multiply two numbers'''
    return a*b
```

```python
result=multiply.invoke({'a':2,'b':3})
result
```

```python
## Three attributes of a tool

print(multiply.name)
print(multiply.args)
print(multiply.description)
```

```python
## When you send this tool to LLM. What LLM sees is given below..

print(multiply.args_schema.model_json_schema())

```



This is what LLM sees

This is the schema of our tool. LLM doesn't see the logic rather it sees the schema. We don't send the tool to LLM, rather we send the schema to the LLM.

## Custom tool using Pydantic and StructuredTool

A Structured Tool is a special type of tool where the input follows a structured schema, typically defined using Pydantic.

Earlier we just did type hinting not strict validation. Now there is strict type validation

```python
from langchain_core.tools import StructuredTool
from pydantic import Field,BaseModel


class MultiplyInput(BaseModel):
    a:int = Field(description='The first number to add',required=True)
    b:int = Field(description="The second number to add",required=True)


def multiply(a:int,b:int)->int:
    return a*b
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply,
    description='Multiply two numbers',
    args_schema=MultiplyInput, ## pydantic class we give as input 
    name='multiply'
)
```

```python
result=multiply_tool.invoke({'a':4,'b':7})
print(result)
```

```python
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args_schema)
print(multiply_tool.func)
```

```python
print(multiply_tool.args)
```

## Using BaseTool class

BaseTool is the Abstract Class for all tools in LangChain. It defines the core structure and interface that any tool must follow, whether it is simple one liner or a fully customized function

All other tool types like @tool, StructuredTool are built on top of BaseTool

```python
from typing import Type
Type[BaseModel]
```

```python
from pydantic import BaseModel,Field
from langchain.tools import BaseTool
from typing import Type
class MultiplyInput(BaseModel):
    a:int=Field(description='First number to add',required=True)
    b:int=Field(description='Second number to add',required=True)


class MultiplyTool(BaseTool):
    name:str='Multiply'
    description:str='Multiply two numbers'
    args_schema:Type[BaseModel]=MultiplyInput

    def _run(self,a:int,b:int):
        return a*b


```

```python
multiply_tool=MultiplyTool()

result=multiply_tool.invoke({'a':3,'b':4})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

## Toolkits

A Toolkit is a collection(bundle) of related tools that serve a common purpose, packaged together for convinence and reusability

Think about a Swiss Army knife. It contains:

- Knife
- Screwdriver
- Scissors
- Bottle opener

All of these are separate tools, but they are bundled together because they serve related purposes. A LangChain toolkit works similarly.

## Example: GoogleDriveToolkit

Imagine you want an LLM to work with Google Drive. You don't want to create one giant function: google_drive_everything()

Instead, you have multiple specialized tools:


```python
# GoogleDriveToolkit
#         │
#         ├── GoogleDriveCreateFileTool -> Upload/create a file
#         │
#         ├── GoogleDriveSearchTool -> Find a file based on its name/content 
#         │
#         └── GoogleDriveReadFileTool -> Read the content of a file

# Each tool has a specific job.
```

## Why Toolkit? 

Imagine building an AI assistant that works with Google Drive.

Without a toolkit:

- Create file
- Search file
- Read file
- Delete file
- Move file

...

You would need to manage many individual tools yourself.

A toolkit groups related tools together:

```python
    #          GoogleDriveToolkit
    #                 │
    #    ┌────────────┼────────────┐
    #    ↓            ↓            ↓
    # Create        Search        Read
    #  File          File         File


# This makes the tools:

# 1. easier to organize
# 2. easier to reuse
# 3. easier to provide to an agent
```

## Tool vs Toolkit

Tool: One specific capability. Example: GoogleDriveSearchTool

Toolkit: A collection of related capabilities.

## Custom Toolkit 

```python
from langchain_core.tools import tool

## custom tools
@tool
def add(a:int, b:int)->int:
    '''Add two numbers'''
    return a+b

@tool
def multiply(a:int,b:int)->int:
    '''Multiply two numbers'''
    return a*b

```

```python
## To create a toolkit you create a class
class MathToolKit:
    def get_tools(self):
        return [add,multiply]
```

```python
toolkit=MathToolKit()
tools=toolkit.get_tools()
```

```python
tools
```

```python
for tool in tools:
    print(tool.name,'=>',tool.description)
```
