## Foundation Models


Types of AI Models that are very large scaled models, trained on very huge amount of data and you need powerful hardwares to train them.

Unlike ML Models, these models are generalized models and not task specific models. Because their architecture is huge and have billions of parameters we gave it a lot of data (whole internet data)


Types of Foundation Models:

1-LLM (Large Lang models)

2-LMM (Large multimodal models)

LLMs can do task like text generation, text classification, sentiment analysis and Q/A (Many tasks not a specific task)

Similarly LMMs work with audio,video and Images

## What is a Framework in GenAI ?

A framework is a ready-made set of tools, functions, and rules that helps you build complex
things easily without starting from scratch.

Ex- A kitchen full of utensils, oven, recipes, and ingredients — so you don’t have to farm wheat,
make flour, or build a stove to cook a pizza.

In GenAI:

• You want to build apps using LLMs (like chatbots, search agents, etc.)

## Frameworks like LangChain, LangGraph, or CrewAI help you:

o Connect to LLMs (OpenAI, Claude, etc.)

o Add memory

o Add tools (e.g., Google Search, database, file reader)

o Use logic and reasoning

## What is an Agent Framework?

An Agent Framework is a framework built specifically to help you make "AI Agents".

Agent = LLM + Memory + Tools + Reasoning

Agents can:

• Think step-by-step (reasoning)

• Use tools (e.g., calculator, search, custom API)

• Make decisions (what to do next)


Agent Frameworks help you:

• Connect the LLM to tools

• Create logic (like: If answer not found → try Google)

• Handle memory, feedback loops, retrying

## Popular Frameworks:

1. LangChain- CrewAI AutoGPT General LLM apps + Agents + Chains

3. CrewAI- Multi-agent systems (team of agents)
